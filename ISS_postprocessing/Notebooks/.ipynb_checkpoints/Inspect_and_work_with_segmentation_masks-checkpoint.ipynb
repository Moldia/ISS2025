# Segmentation Mask Visualization and Contour Export

This notebook demonstrates how to work with segmentation masks generated by **Cellpose** or **Stardist** in the ISS postprocessing pipeline.  

### Workflow

1. **Load Segmentation Mask**  
   The sparse `.npz` mask (e.g., `cellpose_expanded.npz`) is loaded and converted into a dense label image.  

2. **Load Corresponding Raw Image**  
   The raw **DAPI/fluorescence channel** from preprocessing is loaded  for visualization.  

3. **Crop Region of Interest**  
   A subregion of the image is extracted for inspection (e.g., `4000:8000 × 4000:8000`) to focus on a smaller field of view.  

4. **Normalize and Brighten Image**  
   The raw intensities are scaled to `[0, 1]` and enhanced for better contrast.  

5. **Overlay Segmentation Boundaries**  
   Cell boundaries from the label mask are overlaid on the brightened raw image using `skimage.segmentation.mark_boundaries`.  

6. **Generate Contour Mask**  
   - Outer boundaries are extracted from the label mask with `find_boundaries(mode="outer")`.  
   - Boundaries are thickened via **binary dilation** to make them more visible.  
   - The result is saved as an 8-bit TIFF file (white contours on black background).
     

7. **Save to Segmentation Folder**  
   The contour mask is written to:
   `<input_dir>/<region>/postprocessing/segmentation/<region>_<segmentation_method>_contour_mask.tif`
    ensuring consistency with the pipeline’s folder structure.



### Imports

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from tifffile import imread, imwrite
from skimage.segmentation import mark_boundaries, find_boundaries
from scipy.ndimage import binary_dilation
from skimage.morphology import disk
```

### Input Parameters

**`input_dir`** *(str)*: Path to the parent directory containing the **preprocessed region folders**  
  (e.g., `/R1/`, `/R2/`, …). These region folders are created automatically during preprocessing.  

**`region`** *(str)*: Region identifier to process (e.g., `"R1"`).  

**`DAPI_ch`** *(int, default = 4)*: Index of the **DAPI channel** in the raw images (0-based). Example: `DAPI_ch = 4` selects `Cycle1_ch4.tif`.  

**`segmentation_method`** *(str)*: Which segmentation mask to use: `"cellpose"` or `"stardist"`. Must match the mask file you previously saved during segmentation.

> **Note:** In this notebook, **only one region is processed at a time**.  
> The variable `region` must be a **single string** (e.g., `"R1"`), not a list of regions as in earlier pipeline steps.  



```python
input_dir = '/path/to/regions/'
region = 'R1'
DAPI_ch = 4
segmentation_method = 'cellpose' 
```

```python
# --- Load sparse Stardist/Cellpose mask (COO format) ---

segmentation_dir = Path(input_dir) / region / "postprocessing" / "segmentation" 
mask_file = Path(segmentation_dir) / f"{region}_{segmentation_method}_expanded.npz"

# Load sparse npz
data = np.load(mask_file)
row, col, vals, shape = data["row"], data["col"], data["data"], data["shape"]

# Build sparse matrix and convert to dense label image
mask_sparse = coo_matrix((vals, (row, col)), shape=shape)
labels = mask_sparse.toarray().astype(np.int32)

print(f"Loaded segmentation mask from: {mask_file}")
print(f" - Shape: {labels.shape}")
print(f" - Unique labels: {len(np.unique(labels))}")
```

```python
# Build path to raw DAPI / fluorescence image
DAPI_file = Path(input_dir) / region / "preprocessing" / "Cycle1" / "3_stitched" / f"Cycle1_ch{int(DAPI_ch)}.tif"

# Load image
image_wh = imread(DAPI_file)

print("Loaded image:", DAPI_file)
print("Shape:", image_wh.shape, "dtype:", image_wh.dtype)

```

```python
# Define function for segmentation overlay
def plot_segmentation_overlay(
    image, labels, 
    crop_coords=None, 
    brightness_factor=4, 
    figsize=(8, 8), dpi=300, 
    title="Segmentation Overlay (Brightened)"
):
    # Check crop
    if crop_coords:
        y_start, y_end, x_start, x_end = crop_coords

        if y_end > image.shape[0] or x_end > image.shape[1]:
            print(f"Crop larger than image. Showing full image instead.")
        else:
            # Valid crop → apply it
            image = image[y_start:y_end, x_start:x_end]
            labels = labels[y_start:y_end, x_start:x_end]
            print(f"Cropped image shape: {image.shape}")
    else:
        print(f"Using full image, shape: {image.shape}")

    # Normalize and brighten
    image_min, image_max = image.min(), image.max()
    image_range = image_max - image_min if image_max > image_min else 1
    image_norm = (image - image_min) / image_range
    image_bright = np.clip(image_norm * brightness_factor, 0, 1)

    # Overlay boundaries
    overlay = mark_boundaries(
        image_bright, labels, 
        color=(1, 1, 0), 
        mode="outer", 
        background_label=0
    )

    # Plot
    plt.figure(figsize=figsize, dpi=dpi)
    plt.imshow(overlay)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
# Plot segmentation overlay
crop_coords = (4000, 8000, 4000, 8000)  # (y_start, y_end, x_start, x_end)
plot_segmentation_overlay(image_wh, labels, crop_coords=crop_coords)

```

```python
# Define functino to extract contour mask
def extract_contour_mask(labels, thickness=2):
    
    # Step 1: find outer boundaries
    contour = find_boundaries(labels, mode="outer")

    # Step 2: thicken boundaries
    thick_contour = binary_dilation(contour, structure=disk(thickness))

    # Step 3: convert to 8-bit
    contour_image = (thick_contour.astype(np.uint8)) * 255
    return contour_image
```

```python
# Extract contour mask
contour_image = extract_contour_mask(labels, thickness=3)
```

```python
# Output file path
out_path = segmentation_dir / f"{region}_{segmentation_method}_contour_mask.tif"

# Save as TIFF
imwrite(out_path, contour_image.astype(np.uint8))
print(f"Contour mask saved to: {out_path}")
```
