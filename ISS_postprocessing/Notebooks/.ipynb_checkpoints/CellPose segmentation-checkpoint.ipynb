## Cell segmentation using Cellpose

In this step we build a **segmentation mask** from the DAPI signal.  
For each region, the script searches for the stitched DAPI image in:

`/preprocessing/Cycle1/3_stitched/Cycle1_ch4.tif`

By default, channel **4** is used for DAPI. Update `DAPI_ch` if your data uses a different channel.  

Segmentation methods are implemented in `ISS_postprocessing.segmentation`.  
Both **StarDist** and **Cellpose** are available. Custom segmentation methods can also be used, but must be run outside this notebook.  

In this tutorial, segmentation is performed with **Cellpose**.  

### Function parameters

`input_dir` *(str)*: Path to the parent directory containing the preprocessed region folders (e.g. `/R1/`, `/R2/`, …). These region folders are created automatically during preprocessing.  

`region` *(list of str)*: Region identifiers to process.  

`DAPI_ch` *(int, default=4)*: Index of the DAPI channel (0-based).  

`diameter` *(float or int, optional)*: Approximate object diameter for Cellpose (in pixels). If not set, Cellpose will estimate automatically.  

`expanded_distance` *(int, default=20)*: Number of pixels by which each segmented nucleus is expanded.  


> **Note:**  
> You need to specify the names of the regions you want to post-process.  
> Multiple regions can be processed in one run, and they will all be analyzed using the same parameters.  
> In most cases this is fine, but there may be situations where individual regions need to be treated differently.
